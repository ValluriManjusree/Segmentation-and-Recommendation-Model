```python
def create_segment(row):
    if row["TenureMonths"] >= df["TenureMonths"].median() and row["SupportTickets"] <= df["SupportTickets"].median():
        return "Loyal & Stable"

    elif row["TenureMonths"] < df["TenureMonths"].median() and row["SupportTickets"] > df["SupportTickets"].median():
        return "New & High Support"

    elif row["TenureMonths"] < df["TenureMonths"].median():
        return "Newer Customers"

    else:
        return "Established Customers"


df["Segment"] = df.apply(create_segment, axis=1)

df[[
    "CustomerID",
    "TenureMonths",
    "MonthlyCharges",
    "SupportTickets",
    "ContractType",
    "Segment",
    "Churn"
]]
```

```python
segment_summary = df.groupby("Segment").agg(
    Customers=("CustomerID", "count"),
    Avg_Tenure=("TenureMonths", "mean"),
    Avg_Monthly_Charges=("MonthlyCharges", "mean"),
    Avg_Support_Tickets=("SupportTickets", "mean"),
    Churned_Customers=("Churn", lambda x: (x == "Yes").sum())
).reset_index()

segment_summary
```

```python
segment_summary["Churn_Rate"] = (
    segment_summary["Churned_Customers"]
    / segment_summary["Customers"]
    * 100
).round(1)

segment_summary
```

```python
tenure_median = df["TenureMonths"].median()
tickets_median = df["SupportTickets"].median()

df["Risk_Tenure"] = np.where(
    df["TenureMonths"] < tenure_median, 2, 0
)

df["Risk_Contract"] = np.where(
    df["ContractType"] == "Month-to-Month", 2, 0
)

df["Risk_Support"] = np.where(
    df["SupportTickets"] > tickets_median, 1, 0
)

df["Churn_Risk_Score"] = (
    df["Risk_Tenure"]
    + df["Risk_Contract"]
    + df["Risk_Support"]
)

df[[
    "CustomerID",
    "TenureMonths",
    "ContractType",
    "SupportTickets",
    "Risk_Tenure",
    "Risk_Contract",
    "Risk_Support",
    "Churn_Risk_Score",
    "Churn"
]]
```

```python
def risk_level(score):
    if score >= 4:
        return "High Risk"
    elif score >= 2:
        return "Medium Risk"
    else:
        return "Low Risk"

df["Risk_Level"] = df["Churn_Risk_Score"].apply(risk_level)

df[[
    "CustomerID",
    "Churn_Risk_Score",
    "Risk_Level",
    "Churn"
]]
```

```python
risk_summary = df.groupby("Risk_Level").agg(
    Customers=("CustomerID", "count"),
    Avg_Risk_Score=("Churn_Risk_Score", "mean"),
    Churned_Customers=("Churn", lambda x: (x == "Yes").sum())
).reset_index()

risk_summary
```

# Customer Segmentation and Churn-Risk Recommendation Memo

## Objective

The objective of this analysis is to segment customers based on account and behavioral characteristics and create a transparent, rule-based churn-risk score that can be used to prioritize customer-retention actions.

## Segmentation Approach

Customers were segmented using tenure and support-ticket activity. The approach uses simple, reproducible rules so that each segment can be easily understood and explained.

The segments are:

* **Loyal & Stable:** Longer-tenure customers with relatively fewer support tickets.
* **New & High Support:** Newer customers with relatively higher support activity.
* **Newer Customers:** Customers with shorter tenure who may require additional engagement.
* **Established Customers:** Longer-tenure customers who do not fall into the stable segment.

## Churn-Risk Scoring

A rule-based score from 0 to 5 was created using three observable signals:

* Short tenure: +2 points
* Month-to-month contract: +2 points
* Higher-than-median support tickets: +1 point

Risk levels were assigned as:

* **0–1:** Low Risk
* **2–3:** Medium Risk
* **4–5:** High Risk

The score is intentionally transparent so that a business user can understand why a customer received a particular risk level.

## Recommendations

Customers classified as higher risk can be prioritized for retention review. For customers with short tenure and high support activity, the business can consider onboarding assistance and proactive support. Customers on month-to-month contracts can be considered for appropriate engagement or contract-retention initiatives.

Lower-risk customers can continue through regular customer-success activities rather than receiving the same level of intervention.

## Validation

The model was checked for missing risk levels, valid score ranges, and correct application of the individual scoring rules. The dataset contained no missing values, and the scoring logic was applied consistently across the customer records.

Observed churn was used only for validation and interpretation; it was not used as an input to calculate the risk score.

## Limitations

This analysis uses a small sample of 15 customers, so the results should not be generalized to a larger customer population without additional data. The risk score is a prioritization tool rather than a prediction of whether an individual customer will churn.

Further validation on a larger historical dataset would be required before using the score for automated customer decisions.


# Final Project Summary

This project developed an explainable customer segmentation and churn-risk prioritization model using customer account and behavioral data.

Customers were segmented using tenure and support-ticket activity. A transparent churn-risk score was created using short tenure, month-to-month contracts, and higher-than-median support activity as risk signals.

The model provides a simple way to prioritize customers for retention review while keeping the reasoning behind each score understandable to business users.

The analysis also included validation of score ranges, risk-level assignment, and edge cases. Observed churn was used for validation rather than as an input to the risk score.

Because the dataset contains only 15 customers, the results are intended for demonstration and decision-support purposes. A larger historical dataset would be required for stronger validation and production use.
